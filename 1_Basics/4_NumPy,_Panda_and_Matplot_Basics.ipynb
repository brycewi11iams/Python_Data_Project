```python
!pip install datasets
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup (not needed for all problems but good to do anyways)
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
job_country = df.job_country.value_counts().head(5)
```

```python
plt.barh(job_country.index, job_country, color='orange')
```

```python
plt.barh(df.job_title_short, df.salary_year_avg)
```

```python
group = df.groupby('job_title_short')['salary_year_avg'].mean().sort_values()
```

```python
plt.barh(group.index, group.values)
plt.xlabel('Average Salary')
plt.title('Average Salary by Job Title')
```

```python
df['job_posted_month'] = df['job_posted_date'].dt.month
```

```python
month = df.job_posted_month.value_counts()
```

```python
salary_avg = df.groupby('job_posted_month')['salary_year_avg'].mean()
plt.plot(salary_avg.index, salary_avg.values, color='red', label='Average Salary')
plt.xlabel('Month')
plt.ylabel('Average Salary')
plt.title('Average Salary by Job Month')
plt.legend()
plt.show()
```
