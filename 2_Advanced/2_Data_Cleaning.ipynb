```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot

dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
median_salary_year = df['salary_year_avg'].median()
median_salary_hour = df['salary_hour_avg'].median()
```

```python
df_filled = df

df_filled['salary_year_avg'] = df_filled['salary_year_avg'].fillna(median_salary_year)
df_filled['salary_hour_avg'] = df_filled['salary_hour_avg'].fillna(median_salary_year)
```

```python
df_unique = df_filled
df_unique = df_unique.drop_duplicates()
```

```python
df_unique = df_unique.drop_duplicates(subset=['job_title', 'company_name'])

print('Length of oringal df:           ', len(df_filled))
print('Length of dropped duplicates df:', len(df_unique))
print('Rows Dropped                    ', len(df_filled) - len(df_unique))
```

```python
df_dropped = df.dropna(subset=['salary_year_avg'])

print('Length of oringal df:', len(df_filled))
print('rows after           ', len(df_dropped))
```
